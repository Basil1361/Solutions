```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.firefox.options import Options
import time
import random
import pandas as pd

options = Options()
options.set_preference("general.useragent.override", "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36")
driver = webdriver.Firefox(options=options)
driver.get("https://www.brickz.my/transactions/residential/selangor/shah-alam/")
element = driver.find_element(By.XPATH, '/html/body/div[1]/div[2]/article/div[2]/nav/a[1]')
element.click()
time.sleep(3)

Name = []
Tenure_Type = []
Median_Price_Psf = [] 
Median_Price = []
Filed_Transactions = []

while True:
    rows = driver.find_elements(By.XPATH, '//table/tbody/tr')
    for row in rows:
        cells = row.find_elements(By.TAG_NAME, 'td')
        if len(cells) >= 5:
            Name.append(cells[0].text)
            Tenure_Type.append(cells[1].text)
            Median_Price_Psf.append(cells[2].text)
            Median_Price.append(cells[3].text)
            Filed_Transactions.append(cells[4].text)
    try:
        next_button = driver.find_element(By.LINK_TEXT, '»')
        next_button.click()
        time.sleep(random.uniform(8, 15))
        page_count += 1
    except: 
        break

driver.quit()
df = pd.DataFrame({'Name': Name, 'Tenure_Type': Tenure_Type, 'Median_Price_Psf': Median_Price_Psf, 
                   'Median_Price': Median_Price, 'Filed_Transactions': Filed_Transactions})

df.to_csv('House_Data.csv', index=False)
```
